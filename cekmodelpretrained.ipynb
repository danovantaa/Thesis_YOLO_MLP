```python
from ultralytics import YOLO

# Load model
model = YOLO("yolov8l.pt")

# Menampilkan setiap layer
for i, layer in enumerate(model.model.model):
    print(f"Layer {i}:")
    print(layer)
    print("-" * 50)
```
